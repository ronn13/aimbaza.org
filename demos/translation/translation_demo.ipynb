# MbazaNLP Translation Demo

**English ↔ Kinyarwanda machine translation** using MbazaNLP's fine-tuned NLLB-200 models.

Three domain-specific models are available:

| Model | HuggingFace ID | Best for |
|-------|---------------|----------|
| General | `mbazaNLP/Nllb_finetuned_general_en_kin` | Everyday text, mixed domains |
| Education | `mbazaNLP/Nllb_finetuned_education_en_kin` | Academic, instructional content |
| Tourism | `mbazaNLP/Nllb_finetuned_tourism_en_kin` | Travel, hospitality, cultural content |

All models are licensed under Apache 2.0 and freely available on HuggingFace.

## Setup

Install dependencies (only needed once):

```python
# %pip install torch transformers sentencepiece sacremoses
```

```python
from transformers import pipeline

MODELS = {
    "general":   "mbazaNLP/Nllb_finetuned_general_en_kin",
    "education": "mbazaNLP/Nllb_finetuned_education_en_kin",
    "tourism":   "mbazaNLP/Nllb_finetuned_tourism_en_kin",
}

def load_model(domain="general", src="eng_Latn", tgt="kin_Latn"):
    return pipeline("translation", model=MODELS[domain], src_lang=src, tgt_lang=tgt)

def translate(translator, text, max_length=512):
    return translator(text, max_length=max_length)[0]["translation_text"]
```

## 1. English → Kinyarwanda (general domain)

```python
translator_general = load_model("general", src="eng_Latn", tgt="kin_Latn")

sentences_en = [
    "Hello, how are you?",
    "My name is Claude and I am learning Kinyarwanda.",
    "Rwanda is known for its beautiful hills and lakes.",
    "Open-source AI can help communities build tools in their own language.",
]

for s in sentences_en:
    print(f"EN: {s}")
    print(f"RW: {translate(translator_general, s)}")
    print()
```

## 2. Kinyarwanda → English (general domain)

```python
translator_kin_en = load_model("general", src="kin_Latn", tgt="eng_Latn")

sentences_kin = [
    "Muraho, amakuru?",
    "Amazina yanjye ni Claude.",
    "Rwanda ifite imisozi miremire n'ibirunga bikomeye.",
]

for s in sentences_kin:
    print(f"RW: {s}")
    print(f"EN: {translate(translator_kin_en, s)}")
    print()
```

## 3. Domain comparison

The same sentence translated with each domain model. Domain-adapted models outperform the general model on in-domain text.

```python
edu_sentence = "The student must submit their assignment before the deadline."
tour_sentence = "I would like to book a room near the national park with a lake view."

# Load all three models
models = {domain: load_model(domain) for domain in MODELS}

print("=== Education sentence ===")
print(f"EN: {edu_sentence}")
for domain, t in models.items():
    print(f"[{domain:10s}] {translate(t, edu_sentence)}")

print()
print("=== Tourism sentence ===")
print(f"EN: {tour_sentence}")
for domain, t in models.items():
    print(f"[{domain:10s}] {translate(t, tour_sentence)}")
```

## 4. Batch translation

```python
batch = [
    "Welcome to Rwanda.",
    "Please fasten your seatbelt.",
    "The conference starts at nine in the morning.",
    "Thank you for your contribution.",
]

results = translator_general(batch, max_length=512)
for src, res in zip(batch, results):
    print(f"EN: {src}")
    print(f"RW: {res['translation_text']}")
    print()
```

## 5. Translate your own text

```python
my_text = "Type your sentence here."
my_domain = "general"   # general | education | tourism
my_src = "eng_Latn"     # eng_Latn | kin_Latn
my_tgt = "kin_Latn"     # eng_Latn | kin_Latn

t = load_model(my_domain, src=my_src, tgt=my_tgt)
print(translate(t, my_text))
```
